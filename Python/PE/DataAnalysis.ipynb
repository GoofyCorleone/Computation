```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

names1 =['Malus_Tradicional_1/{}.txt'.format(i) for i in np.arange(0,380,20)]
names2 =['Malus_Tradicional_2/{}.txt'.format(i) for i in np.arange(0,380,20)]
names3 =['Malus_Tradicional_3/{}.txt'.format(i) for i in np.arange(0,380,20)]


def gettinData(route):
    Data = pd.read_csv(route, header=19, delimiter='\t')
    Counts = Data['Counts per Bin '].values
    return Counts

def getCounts(names):
    conteo = []
    # RuidoAmb = gettinData('Malus_Tradicional/RuidoAmb.txt')
    #RuidoAmb = gettinData('Malus_Tradicional_1/0.txt')
    for name in names:
        vals = gettinData(name)
        # vals = vals[ vals > ruido ]
        #vals = vals[ vals > max(RuidoAmb) ]
        #vals = vals - RuidoAmb
        if vals.size == 0:
            val = 0
        else:
            val = vals.mean()
        conteo.append(val)
    conteo = conteo /max(conteo)
    return conteo

angles = np.arange(0,380,20)
ang_arr = np.linspace(0,2*np.pi,300)
Itheta = (np.cos(ang_arr + 0) )**2

conteo1 = getCounts(names=names1)
conteo2 = getCounts(names=names2)
conteo3 = getCounts(names=names3)
conteo4 = ( conteo1 + conteo2 + conteo3 ) / 3

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0,0].scatter(angles,conteo1)
axs[0,0].plot(ang_arr * 180 / np.pi, Itheta, '--')

axs[0,1].scatter(angles,conteo2)
axs[0,1].plot(ang_arr * 180 / np.pi, Itheta, '--')

axs[1,0].scatter(angles,conteo3)
axs[1,0].plot(ang_arr * 180 / np.pi, Itheta, '--')

axs[1,1].scatter(angles,conteo4)
axs[1,1].plot(ang_arr * 180 / np.pi, Itheta, '--')

plt.show()
```
